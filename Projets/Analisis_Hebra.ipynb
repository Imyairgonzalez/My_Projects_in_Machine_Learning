```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/Users/eduardoyair/Downloads/analisishebra.csv')
df.fillna(0)   
```

```python
x = df['totalcasein']
y = df['ACIDITY']
plt.plot(x,y,'o',color = 'purple')
plt.xlabel('Casein in the Raw mil ')
```

```python
casein = pd.DataFrame(
    {
        'ones': 1,
        'totalcasein': df['totalcasein']
    }
)
vector_casein = casein.values
print(vector_casein)
```

```python
acidity = df['ACIDITY']
vector_acidity = acidity.values
np.shape(vector_acidity)
```

In this part i need to aproximate the value of thetha1 and thetha2, remember that: 


Y = X A + E
A = (X.T @ X)^-1 @ X.T @ Y

```python
predictor = (np.linalg.inv(vector_casein.T @ vector_casein)) @ (vector_casein.T @ vector_acidity)
print(predictor)
```

The equation of the regression for least squares is : 

Y = thetha1 * x + thetha2

```python
print(f'Y = {predictor[0]}x  {predictor[1]}')
```
