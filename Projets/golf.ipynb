## Bayes Ingenuo

### El algoritmo de Bayes ingenuo, utiliza la probabilidad de observar valores predictores, dado un resultado, para estimar lo que realmente interesa: La probabilidad de observar el resultado $Y = i$, dado un conjunto de valores predictores.

### Probabilidad Condicional

### La probabilidad de observar algun evento (digamos, $X=i$) dado algún otro evento (digamos ,$Y=i$.Se escribe $P(X_i | Y_i)$
\[
 $P(A|B) = \frac{P(A \cap B)}{P(B)}$
\]
### Probabilidad a posteriori
### Probabilidad de un resultado despues de que se haya incorporado la información de la predictora(en contraste con la probabilidad previa de los resltados, sin tener en cuenta la información de la predictora)
\[
$ P(Y_i | X_1, X_2, X_3,...)   $
\]

### Al desarrollar la formula hasta un caso p.
\[
$ P(Y_i | X_1, X_2, X_3,...,X_p) = \frac{P(Y=0) P(X_1,..X_p | Y =i)}{P(Y=0) P(X1_...X_p | Y=O)+ P(Y=1)P(X1_...X_p | Y=1)}   $
\]
