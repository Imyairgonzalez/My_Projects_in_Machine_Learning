```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats 
```

```python
x = stats.norm(5,2.3)
x.mean()
x.std()
```

```python
print(f'Valor esperado es: {x.mean()}')
```

```python
x.pdf([i for i in range(40)])
```

```python
x.cdf([i for i in range(50)])
```

```python
x.interval(0.99)
```

```python
#Valor esperado
x.expect()
```

```python
x_min, x_max = x.interval(0.99)
x_domain = np.linspace(x_min,x_max,100)
y = ((x_domain - x.mean())/x.std())**2
```

```python
fig, ax = plt.subplots()
ax.plot(x_domain,x.cdf(x_domain),"-",label="Acumulada", color="g")
ax.plot(x_domain,x.pdf(x_domain),"-",label="Densidad",color="b")
plt.grid()
plt.legend()
```

### Pasemos de una normal(n,o) a una chi

```python
x = stats.norm(5,2.3)
x_muestra = x.rvs(200000)
muestra_z = (x_muestra - x.mean()) / x.std()
muestra_z2 = (muestra_z)**2
plt.hist(muestra_z2, bins=50, color='skyblue', label='Simulación erica ($Z^2$)')
```
