We have a solve problem, in this case we have values of age and height. Me need to predice the weight use with base the height. We use the least squares regression a problem very famous in Lineal Algebra.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

table = {173:81, 171:72, 189:96, 181:94}
print('Height  Weight')
print(f'{pd.Series(table)}\n')
```

```python
x = [173,171,189,181]
y = [81,72,96,94]
plt.plot(x,y,'o')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.grid()
```

Remember the equation of lineal fuction Y = XA + E where Y is the column vector predictor and A is the column vector that have the Height and E is the sum of the errors(the diference betwen the (x-f(x)) at square.

¿And how to solve A?

To solve A is necessary use the Property:

A = (X^T*X)^-1 * X^T*Y


and 

E = E^T * E

Y = column vector predictor (weight)


X = column vector features(height)

```python
x = np.array([[1,173],[1,171],[1,189],[1,181]])
y = np.array([[81],[72],[96],[94]])

s = np.linalg.inv((x.T)@x)
a = ((x.T)@ y)
c = (s @ a)
b = c[0]
m = c[1]
```

Finally we define the fuction value_missing that predice the Weight use the Height use the least squares regression.

```python
def value_missing(m,b,e):
    print(f'¿Cual es el peso que podria tener una persona si tiene {e} cm de Altura?')
    return b + m*e

print(value_missing(m,b,172))
```

Remember tha the equation predictory is the best straight line  shuch that the sum of the distance between (x-f(x)) at square is the least of all.
